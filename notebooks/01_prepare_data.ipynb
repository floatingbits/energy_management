```python
import pandas as pd

path = '../data/real_pv/br-pvgen/'
metadata_df = pd.read_csv(path + "metadata.csv")

# print(metadata_df)
inv_cols = ["datetime", "ps_id","total_active_power_w"]
meteo_cols = ["ambient_temperature_celsius", "panel_temperature_celsius", "ghi_irradiance_wm2", "poa_irradiance_wm2"]
joined_cols = meteo_cols + inv_cols
meteo_cols += ["datetime", "ps_id"]
result = None
for pv_id in metadata_df['id'].unique():
    inv_filename = path + 'inverter/' + pv_id + '.csv'
    inv_df = pd.read_csv(inv_filename)
    inv_df["datetime"] = pd.to_datetime(inv_df["datetime"])
    inv_df = inv_df.groupby(["datetime", "ps_id"], as_index=False).agg(total_active_power_w=("total_active_power_w", "sum"))
    inv_df = inv_df[inv_cols]
    #print(pv_id, " inv ", inv_df["datetime"].min(), " - ", inv_df["datetime"].max())
    inv_df = inv_df.set_index(["datetime", "ps_id"])
    
    
    meteo_filename = path + 'meteorological/' + pv_id + '.csv'
    meteo_df = pd.read_csv(meteo_filename)
    meteo_df["datetime"] = pd.to_datetime(meteo_df["datetime"])
    #print(pv_id, " meteo ", meteo_df["datetime"].min(), " - ", meteo_df["datetime"].max())
    meteo_df = meteo_df[meteo_cols]
    meteo_df = meteo_df.set_index(["datetime", "ps_id"])
    
    common_df = inv_df.join(meteo_df)
          
    if result is None:
        result = common_df
    else:
        result = pd.concat([result, common_df])
    



result
    
    
```

```python
result = result.reset_index()
result = result.dropna()
```
