```python
import importlib
import pandas as pd
import shared.br_pv_helpers as br_pv
importlib.reload(br_pv)
# using non-checked-in data from https://zenodo.org/records/21511487
metadata_df = br_pv.get_metadata_df()
best_locations_df = br_pv.get_best_locations_df()
#best_locations_df
```

```python
# Export Metadata with Best Locations for fixtures.
best_locations_df = br_pv.get_best_locations_df()
result_json_strings = []
for ps_id in best_locations_df.index.unique():
    area = br_pv.calc_area_m2(metadata_df, ps_id)
    lat = best_locations_df.loc[ps_id]['lat']
    lon = best_locations_df.loc[ps_id]['lon']
    ps_metadata = metadata_df.loc[ps_id]
    power_kw = ps_metadata['nominal_power_mw'] * 1000
    efficiency = ps_metadata['panel_efficiency_percentage']/100
    result_json_strings.append(
"""{{
  "name": "BR-PV {ps_id}",
  "asset_type": "solar",
  "installed_power_kw": {power_kw},
  "latitude": {lat:.2f},
  "longitude": {lon:.2f},
  "status": "ACTIVE",
  "configuration": {{
    "panel_area_m2": {area:.2f},
    "efficiency": {efficiency:.4f},
    "tilt_deg": {tilt:.2f},
    "azimuth_deg": 0
  }}
}}""".format(ps_id=ps_id, lat=lat, lon=lon, area=area, power_kw=power_kw, efficiency=efficiency, tilt=abs(lat))
    )

print(",\n".join(result_json_strings))
```
